# Food Recall Root-Cause Intelligence: Exploratory Data Analysis of FDA Food Enforcement Reports

## 1. Project Overview

Food recalls protect consumers when a product may be unsafe, but each recall also costs a food company money, product and trust. Recalls can happen because of undeclared allergens, bacterial contamination, foreign material, labeling mistakes, or problems in processing and temperature control. Analyzing recall data helps identify which causes and products appear most often and how recalls are handled.

This project is a data analysis project that explores food recall records using Exploratory Data Analysis (EDA), data cleaning, and visualization techniques. It uses the Food Enforcement Reports published by the U.S. Food and Drug Administration (FDA) through openFDA. The analysis is framed for the Global Food Safety and Quality Assurance team of Nestlé. The dataset is public, industry-wide FDA data and was not provided by Nestlé.

The project focuses on identifying patterns in recall causes, product categories, firms, states, notification channels, recall status and the time taken by the FDA to classify a recall. It also examines data quality so that a reliable, cleaned dataset is available for further analysis.

The primary goal is to transform raw, manually entered recall data into meaningful insights that can support preventive food safety decisions.

## 2. Problem Statement

Food recalls cost manufacturers lost product, regulatory exposure and brand damage. When recall records are analyzed without systematic processing, it is difficult to identify which causes and products lead to the most recalls and how quickly recalls are handled.

The raw FDA recall records are entered manually and are inconsistent. The reason for recall and the product description are free text, dates are stored as text and can be missing or invalid, state values are abbreviations, and many fields are blank or marked "Unknown". These data quality issues can affect the reliability of any analysis and chart built on the data.

Therefore, a systematic data analysis approach is required to clean, validate, organize and explore the food recall records. This project addresses the problem by applying Python-based data preprocessing and visualization techniques to answer the question: **which root causes and product categories lead to the most recalls, and how do recall status and classification time differ across them?**

## 3. Project Objectives

The major objectives of this project are:

1. **Data Acquisition and Understanding:** Load the raw FDA food recall data and examine its structure, dimensions, column names, data types and descriptive statistics.

2. **Data Quality Assessment:** Identify missing values, duplicate rows, duplicated recall numbers, and inconsistent categories and dates.

3. **Data Cleaning and Standardization:** Standardize column names, convert date columns into datetime format, remove extra spaces, map state abbreviations to full state names, and handle missing values with clear labels.

4. **Feature Engineering:** Create new analytical attributes: classification days, recall year, root cause, product category and a timely classification indicator.

5. **Exploratory Data Analysis:** Analyze recall frequency across product categories, root causes, firms, states and notification channels, and examine recall status and classification.

6. **Comparative Data Visualization:** Compare related variables using clear charts to communicate patterns and differences.

7. **Data Quality Validation and Export:** Verify missing values, duplicates, data types and derived features, and save the cleaned dataset as a CSV file.

## 4. Dataset Description

### 4.1 Dataset Information

- **Dataset Name:** FDA Food Enforcement Reports (Food Recalls)
- **Data Source:** U.S. Food and Drug Administration (FDA), downloaded through openFDA (`food/enforcement`)
- **Industry:** Food manufacturing and food safety
- **Data Format:** CSV (Comma-Separated Values), converted from the original JSON download without changing any values
- **Data Type:** Structured tabular data containing categorical variables, free-text fields, dates and identifiers.
- **Initial Dataset Dimensions:** shown in Phase 1 when the file is loaded.

### 4.2 Dataset Overview

The dataset contains one row for each recalled product. Each record provides information about the recall, the product, the reason for the recall, the firm that recalled the product, the firm's state, how customers were notified, the FDA classification, and the dates of the recall.

### 4.3 Description of Important Attributes

| Attribute | Description |
|---|---|
| recall_number | Identifier of the recall record |
| event_id | Event the recall belongs to (one event can cover several products) |
| status | Progress of the recall: Ongoing, Completed or Terminated |
| classification | FDA severity class: Class I (serious harm possible), Class II (temporary or reversible harm), Class III (harm unlikely) |
| recalling_firm | Firm that recalled the product |
| product_description | Free-text description of the recalled product |
| product_quantity | Quantity recalled, stored as text |
| reason_for_recall | Free-text reason for the recall |
| voluntary_mandated | Whether the recall was voluntary or FDA-mandated |
| initial_firm_notification | How the firm first notified customers |
| distribution_pattern | Free-text description of where the product was distributed |
| state | State of the recalling firm, written as an abbreviation |
| recall_initiation_date | Date the firm started the recall |
| center_classification_date | Date the FDA classified the recall |
| termination_date | Date the recall was closed (empty while the recall is ongoing) |
| report_date | Date the record was published |

### 4.4 Dataset Suitability

The dataset is suitable for descriptive and comparative analysis because it contains several categorical attributes and date fields. These variables allow analysis of recall frequency, product-level patterns, recall status, geographical distribution and the time taken to classify a recall.

The reason and product fields are free text, so they must be grouped before they can be compared. The groups created in this project use simple keyword searches and are approximations.

## Phase 4: Data Cleaning and Preprocessing

Create a working copy, standardize column names, convert dates, remove extra spaces, standardize states, handle missing values and check duplicates.

### 4.1 Create a working copy

```python
# Create raw and clean dataset backups
raw_data = data.copy()
clean_data = data.copy()

print("Raw dataset shape:", raw_data.shape)
print("Clean dataset shape:", clean_data.shape)
```

### 4.2 Standardize column names

```python
clean_data.columns = (
    clean_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print("Updated Column Names:\n")

for i, column in enumerate(clean_data.columns, start=1):
    print(f"{i} -> {column}")
```

### 4.3 Convert date columns to datetime
Values that are not valid dates (such as `00000000`) become missing (`NaT`).

```python
clean_data["recall_initiation_date"] = pd.to_datetime(
    clean_data["recall_initiation_date"],
    format="%Y%m%d",
    errors="coerce"
)

clean_data["center_classification_date"] = pd.to_datetime(
    clean_data["center_classification_date"],
    format="%Y%m%d",
    errors="coerce"
)

clean_data["termination_date"] = pd.to_datetime(
    clean_data["termination_date"],
    format="%Y%m%d",
    errors="coerce"
)

clean_data["report_date"] = pd.to_datetime(
    clean_data["report_date"],
    format="%Y%m%d",
    errors="coerce"
)

print(clean_data[
    ["recall_initiation_date", "center_classification_date", "termination_date", "report_date"]
].dtypes)
```

### 4.4 Remove leading and trailing spaces from text values

```python
text_columns = [
    "status",
    "classification",
    "recalling_firm",
    "product_description",
    "product_quantity",
    "reason_for_recall",
    "voluntary_mandated",
    "initial_firm_notification",
    "distribution_pattern",
    "state"
]

for column in text_columns:
    clean_data[column] = (
        clean_data[column]
        .astype("string")
        .str.strip()
    )

print("Text columns cleaned successfully.")
```

### 4.5 Standardize state values

```python
# Convert state abbreviations into full state names

clean_data["state"] = (
    clean_data["state"]
    .str.upper()
    .str.strip()
)

# Dictionary mapping US state abbreviations to full names
state_mapping = {
    "AL": "Alabama",
    "AK": "Alaska",
    "AZ": "Arizona",
    "AR": "Arkansas",
    "CA": "California",
    "CO": "Colorado",
    "CT": "Connecticut",
    "DE": "Delaware",
    "DC": "District of Columbia",
    "FL": "Florida",
    "GA": "Georgia",
    "HI": "Hawaii",
    "ID": "Idaho",
    "IL": "Illinois",
    "IN": "Indiana",
    "IA": "Iowa",
    "KS": "Kansas",
    "KY": "Kentucky",
    "LA": "Louisiana",
    "ME": "Maine",
    "MD": "Maryland",
    "MA": "Massachusetts",
    "MI": "Michigan",
    "MN": "Minnesota",
    "MS": "Mississippi",
    "MO": "Missouri",
    "MT": "Montana",
    "NE": "Nebraska",
    "NV": "Nevada",
    "NH": "New Hampshire",
    "NJ": "New Jersey",
    "NM": "New Mexico",
    "NY": "New York",
    "NC": "North Carolina",
    "ND": "North Dakota",
    "OH": "Ohio",
    "OK": "Oklahoma",
    "OR": "Oregon",
    "PA": "Pennsylvania",
    "PR": "Puerto Rico",
    "RI": "Rhode Island",
    "SC": "South Carolina",
    "SD": "South Dakota",
    "TN": "Tennessee",
    "TX": "Texas",
    "UT": "Utah",
    "VT": "Vermont",
    "VA": "Virginia",
    "WA": "Washington",
    "WV": "West Virginia",
    "WI": "Wisconsin",
    "WY": "Wyoming",
}

clean_data["state"] = clean_data["state"].replace(state_mapping)

clean_data["state"].value_counts().head(10)
```

### 4.6 Handle missing values
Text columns get a clear label. The classification and the dates are **not** filled, because a guessed value would give wrong results. A missing `termination_date` simply means the recall is still ongoing.

```python
clean_data["state"] = clean_data["state"].fillna("Unknown")

clean_data["recalling_firm"] = clean_data["recalling_firm"].fillna("Unknown")

clean_data["product_description"] = clean_data["product_description"].fillna("Not Reported")

clean_data["reason_for_recall"] = clean_data["reason_for_recall"].fillna("Not Reported")

clean_data["product_quantity"] = clean_data["product_quantity"].fillna("Not Reported")

clean_data["voluntary_mandated"] = clean_data["voluntary_mandated"].fillna("Not Reported")

clean_data["initial_firm_notification"] = clean_data["initial_firm_notification"].fillna("Not Reported")

clean_data["distribution_pattern"] = clean_data["distribution_pattern"].fillna("Not Reported")

clean_data.isnull().sum()
```

### 4.7 Check duplicate records
Only rows that are identical in every column are removed. Recall numbers are not used for removal, because the same recall number can legitimately appear for different products.

```python
duplicate_count = clean_data.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
print("Duplicate recall numbers (kept, not removed):", clean_data["recall_number"].duplicated().sum())

# remove duplicates only if they actually exist
if duplicate_count > 0:
    clean_data = clean_data.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")
```

## Phase 5: Feature Engineering

Derive new analytical attributes from existing columns. Every derived value is checked for missing dates and negative durations before it is used.

### 5.1 Classification days
The number of days between the recall start date and the date FDA classified the recall. This is the equivalent of "response duration".

```python
# Days from recall initiation to FDA classification
clean_data["classification_days"] = (
    clean_data["center_classification_date"] - clean_data["recall_initiation_date"]
).dt.days

clean_data[["recall_initiation_date", "center_classification_date", "classification_days"]].head()
```

```python
# Check missing dates and negative values
print("Missing initiation dates:", clean_data["recall_initiation_date"].isnull().sum())
print("Missing classification dates:", clean_data["center_classification_date"].isnull().sum())
print("Missing classification days:", clean_data["classification_days"].isnull().sum())

negative_days = (clean_data["classification_days"] < 0).sum()
print("Negative classification-day records:", negative_days)

# A negative duration is impossible, so it is set to missing
clean_data.loc[clean_data["classification_days"] < 0, "classification_days"] = np.nan
```

### 5.2 Recall year

```python
clean_data["recall_year"] = clean_data["recall_initiation_date"].dt.year

clean_data["recall_year"].value_counts().sort_index()
```

### 5.3 Root cause and product category
The reason and product description are free text, so they are grouped with simple keyword searches (`|` means "or"). A later line replaces the label of an earlier line when both match, so the most specific cause is written last. Anything that matches no keyword stays in the "Other" group. These groups are approximations.

```python
reason = clean_data["reason_for_recall"]

clean_data["root_cause"] = "Other / Unclassified"

clean_data.loc[reason.str.contains("label|mislabel|packag|coding|expiration|date code|misbrand|leak|seal", case=False, na=False), "root_cause"] = "Labeling / Packaging Error"
clean_data.loc[reason.str.contains("pasteuri|underprocess|under-process|temperature|cooling|refrigerat|acidified|sterility|inadequate process|process control|thermal", case=False, na=False), "root_cause"] = "Process / Temperature Control"
clean_data.loc[reason.str.contains("lead|pesticide|toxin|aflatoxin|melamine|chemical|heavy metal|arsenic|cadmium|mercury|sanitizer|cleaning", case=False, na=False), "root_cause"] = "Chemical Contamination"
clean_data.loc[reason.str.contains("metal|glass|plastic|foreign|fragment|wire|stone|rubber|bone|insect", case=False, na=False), "root_cause"] = "Foreign Material"
clean_data.loc[reason.str.contains("botulism|clostridium|cronobacter|staph|bacillus|norovirus|hepatitis|mold|mould|yeast|bacteria|microbial|pathogen|spoilage", case=False, na=False), "root_cause"] = "Other Microbial"
clean_data.loc[reason.str.contains("e. coli|escherichia|stec", case=False, na=False), "root_cause"] = "E. coli"
clean_data.loc[reason.str.contains("salmonella", case=False, na=False), "root_cause"] = "Salmonella"
clean_data.loc[reason.str.contains("listeria|monocytogenes", case=False, na=False), "root_cause"] = "Listeria"
clean_data.loc[reason.str.contains("undeclared|allergen|allergy|milk|wheat|soy|egg|peanut|tree nut|sesame|sulfite|shellfish|fish", case=False, na=False), "root_cause"] = "Undeclared Allergen"
clean_data.loc[reason == "Not Reported", "root_cause"] = "Reason Not Reported"

clean_data["root_cause"].value_counts()
```

```python
description = clean_data["product_description"]

clean_data["product_category"] = "Other Food"

clean_data.loc[description.str.contains("sandwich|ready-to-eat|ready to eat|meal|soup|burrito|pizza|dinner|entree|frozen", case=False, na=False), "product_category"] = "Prepared / Ready-to-Eat"
clean_data.loc[description.str.contains("candy|chocolate|gummy|snack|chip|popcorn|nuts|peanut butter|bars", case=False, na=False), "product_category"] = "Confectionery & Snacks"
clean_data.loc[description.str.contains("juice|drink|beverage|coffee|water|soda|smoothie", case=False, na=False), "product_category"] = "Beverages"
clean_data.loc[description.str.contains("spinach|lettuce|salad|tomato|apple|berry|berries|melon|cantaloupe|sprout|produce|onion|vegetable|fruit", case=False, na=False), "product_category"] = "Fresh Produce"
clean_data.loc[description.str.contains("cookie|bread|cake|cracker|cereal|granola|muffin|bagel|pastry|bakery|flour|pasta|noodle", case=False, na=False), "product_category"] = "Bakery & Cereal"
clean_data.loc[description.str.contains("milk|cheese|yogurt|yoghurt|butter|cream|ice cream|egg|dairy", case=False, na=False), "product_category"] = "Dairy & Eggs"
clean_data.loc[description.str.contains("chicken|beef|pork|turkey|sausage|hot dog|bacon|meat|jerky|poultry", case=False, na=False), "product_category"] = "Meat & Poultry"
clean_data.loc[description.str.contains("salmon|tuna|shrimp|fish|seafood|crab|lobster|oyster|clam|cod|tilapia", case=False, na=False), "product_category"] = "Seafood"
clean_data.loc[description.str.contains("supplement|capsule|vitamin|tablet|protein powder|herbal", case=False, na=False), "product_category"] = "Dietary Supplements"
clean_data.loc[description.str.contains("infant|baby|toddler|formula", case=False, na=False), "product_category"] = "Infant & Baby Food"
clean_data.loc[description == "Not Reported", "product_category"] = "Description Not Reported"

clean_data["product_category"].value_counts()
```

### 5.4 Timely classification indicator
A recall counts as classified on time when FDA classified it within 30 days of the recall start. **The 30-day limit is an assumption made for this project** (the FDA data has no such column). Change the number in the code if your instructor prefers another limit. Recalls with no classification days are marked "Not Available".

```python
# Change this number to use a different limit
timely_limit = 30

clean_data["timely_classification"] = np.where(
    clean_data["classification_days"] <= timely_limit, "Yes", "No"
)

clean_data.loc[clean_data["classification_days"].isnull(), "timely_classification"] = "Not Available"

clean_data["timely_classification"].value_counts()
```

## Phase 7: Comparative Data Visualization

Compare related variables to show differences and trends that are hard to see in tables. Recalls with no classification days ("Not Available") are left out of the timely-classification charts.

### 7.1 Product category vs timely classification

```python
timely_data = clean_data[clean_data["timely_classification"] != "Not Available"]

product_timely = pd.crosstab(
    timely_data["product_category"],
    timely_data["timely_classification"],
    normalize="index"
) * 100

top_products = timely_data["product_category"].value_counts().head(8).index

product_timely = product_timely.loc[top_products]

product_timely.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 7)
)

plt.title("Timely Classification by Product Category", fontsize=15, fontweight="bold")
plt.xlabel("Product Category")
plt.ylabel("Percentage of Recalls (%)")

plt.xticks(rotation=45, ha="right")

plt.legend(title="Classified Within Limit", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

### 7.2 Product category vs recall status

```python
product_status = pd.crosstab(
    clean_data["product_category"],
    clean_data["status"]
)

top_products = clean_data["product_category"].value_counts().head(8).index

product_status = product_status.loc[top_products]

product_status.plot(
    kind="bar",
    stacked=True,
    figsize=(13, 7)
)

plt.title("Recall Status Across Product Categories", fontsize=15, fontweight="bold")
plt.xlabel("Product Category")
plt.ylabel("Number of Recalls")

plt.xticks(rotation=45, ha="right")

plt.legend(title="Recall Status", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

### 7.4 Product category vs classification duration
